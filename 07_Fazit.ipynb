<h2 id="71">7.1. Schlussfolgerungen</h2>

Im Rahmen dieses Projekts wurde ein vollständiger End-to-End-Workflow für ein datengetriebenes Pairs-Trading-Setup umgesetzt. Beginnend bei der automatisierten Datensammlung über ein umfangreiches Preprocessing bis hin zur explorativen Analyse, Modellierung und einem realistischen Backtest mit Portfolio- und Risikologik konnten zentrale Herausforderungen im Umgang mit hochfrequenten Finanzzeitreihen praktisch erfahren werden.

Ein wesentliches Ergebnis ist die Erkenntnis, dass **Datenqualität und Struktur** einen deutlich grösseren Einfluss auf den Projekterfolg haben als die Wahl eines einzelnen Modells. Insbesondere die Behandlung fehlender Daten, zeitlicher Inkonsistenzen und unvollständiger Zeitreihen erwies sich als kritisch für alle nachfolgenden Analysen.

Die explorative Analyse zeigte, dass hohe Korrelation allein nicht ausreicht, um geeignete Aktienpaare zu identifizieren. Erst die Kombination aus Korrelation und **täglicher Cointegration** liefert robuste Kandidaten für Mean-Reversion-basierte Strategien. Der Z-Score des Spreads bestätigte sich dabei als zentrales State-Feature mit klar interpretierbarem Mean-Reversion-Verhalten.

In der Modellierung zeigte sich, dass nichtlineare und ensemblebasierte Modelle (insbesondere XGBoost) die lineare Baseline übertreffen können, wenn sie unter zeitlich konsistenten Bedingungen evaluiert werden.

Die zentrale, am gründlichsten geprüfte Erkenntnis des gesamten Projekts liefert jedoch erst der Backtest in Kapitel 6. Das Modell besitzt eine reale statistische Edge, formal signifikant und robust gegenüber Autokorrelation und Kalibrierungsunsicherheit sowie separat bestätigt auf Tuning- und Holdout-Daten (siehe 6.8). Diese Edge reicht bei der gehandelten Frequenz jedoch nicht aus, um realistische Transaktionskosten zu decken (6.4), und sie schneidet in der Gesamtbilanz sogar schlechter ab als eine simple, klassische Mean-Reversion-Regel ohne ML (6.7), weil das Modell schlicht sehr viel häufiger handelt, ohne dafür einen ausreichend grösseren Ertrag je Position zu erzielen. Jeder naheliegende Hebel, um diese Lücke zu schliessen (Eintritts-Schwelle, Ausstiegsregel, Tageszeit, Risikolimit), wurde einzeln mit Tuning-/Holdout-Disziplin geprüft und hat sie nicht geschlossen (6.4.1, 6.6).

Ein zentrales Lernziel des Projekts wurde damit auf einer deutlich rigoroseren Grundlage erreicht, als es die reine Modellevaluation in Kapitel 5 allein zeigen konnte: Die Entwicklung eines Modells mit guter Metrik-Performance, sogar mit einer statistisch nachweisbaren Edge, ist nicht gleichbedeutend mit der Entwicklung einer wirtschaftlich tragfähigen Handelsstrategie. Statistische Signifikanz eines Signals bedeutet nicht automatisch ökonomische Handelbarkeit.

Darüber hinaus wurde im Verlauf des Projekts deutlich, dass der tatsächliche Arbeitsaufwand und die inhaltliche Tiefe deutlich grösser waren als zu Beginn erwartet. Insbesondere Datensammlung, Preprocessing, explorative Analyse und der Backtest nahmen einen erheblichen Teil der Projektzeit in Anspruch. Diese Erfahrung erwies sich als wertvoll, da sie half, den Umfang der geplanten Bachelorarbeit realistischer einzuschätzen und thematisch klarer einzugrenzen. Das vorliegende Projekt leistete damit auch eine wichtige konzeptionelle Vorbereitung für weiterführende Arbeiten.

Das Projekt ist ausdrücklich als Lern- und Explorationsarbeit im Rahmen des Moduls zu verstehen und erhebt keinen Anspruch auf die Entwicklung einer produktionsreifen oder profitablen Handelsstrategie.


<h2 id="72">7.2. Ausblick</h2>

Das Projekt stellt eine konzeptionelle Grundlage für weiterführende Arbeiten dar, insbesondere im Kontext der geplanten Bachelorarbeit. Für eine Weiterentwicklung ergeben sich mehrere Ansatzpunkte:

- **Datenbasis und Preprocessing:**
  Das verwendete Aktienuniversum weist einen Survivorship Bias auf, da nur S&P-500-Mitglieder berücksichtigt werden die über die gesamte Zeitlänge eine volle Coverage haben. Für Mean-Reversion-Strategien ist dies problematisch und müsste in zukünftigen Arbeiten explizit adressiert werden. Zudem könnten Handelskalender, Corporate Actions und Volumeninformationen sauber integriert werden.

- **Vertiefte Explorative Datenanalyse:**
  Die EDA könnte um sektorbasierte Analysen, Regimewechsel, Intraday-Saisonalitäten oder längere Zeiträume erweitert werden. Insbesondere im Finanzkontext gilt: EDA kann kaum „zu viel“ betrieben werden.

- **Modellierung und Zieldefinition:**
  Anstelle einer rein richtungsbasierten Klassifikation könnten Regressionsansätze, quantile-basierte Modelle oder Reinforcement-Learning-Ansätze untersucht werden. Darüber hinaus besteht erhebliches Potenzial für ein weitergehendes **Hyperparameter-Tuning**, das im Projekt bewusst nur eingeschränkt vorgenommen wurde, um die Rechenkomplexität zu begrenzen.
  Künftig könnten auch adaptive Zieldefinitionen sowie eine explizite Kopplung von Modelloutput und Positionsgrösse untersucht werden. Dabei ist jedoch zu betonen, dass im Rahmen dieses Projekts **kein vollständiges Trading-System** entwickelt oder optimiert werden soll, sondern der Fokus auf der methodischen Modellierung liegt.

- **Trading-Logik und Risikomanagement:**
  Kapitel 6 hat bereits Positionsgrössen, Portfolioaggregation, ein Risiko-Framework sowie eine formale Signifikanzprüfung ergänzt und dabei gezeigt, dass die vorhandene statistische Edge bei der aktuellen Handelsfrequenz nicht kostendeckend ist. Offen bleibt insbesondere, ob eine deutlich günstigere Kostenstruktur (z. B. institutioneller Marktzugang) an diesem Bild etwas ändern würde; das wurde in diesem Projekt bewusst nicht untersucht. Darüber hinaus könnte die Trading-Logik künftig noch stärker an Erkenntnisse aus der **Zeitreihenanalyse der EDA** angepasst werden, etwa durch die Berücksichtigung von Intraday-Saisonalitäten, Volatilitätsregimen oder Mean-Reversion-Stärken zu unterschiedlichen Tageszeiten. Eine solche Erweiterung würde jedoch den Umfang dieses Projekts deutlich überschreiten und ist ausdrücklich nicht Ziel der vorliegenden Arbeit.

- **Bachelorarbeit:**
  In der geplanten Bachelorarbeit soll das Thema **Multi-Asset Statistical Arbitrage** unter Einsatz von **Topological Deep Learning** untersucht werden. Das vorliegende Projekt dient dabei als methodische und konzeptionelle Orientierungshilfe, stellt jedoch keine direkte Vorarbeit dar. Die betrachtete Anlagenklasse und der methodische Fokus werden sich unterscheiden.
