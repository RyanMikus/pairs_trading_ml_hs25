# 1. Einleitung

## 1.1. Projektübersicht

Dieses Projekt wurde im Rahmen des Moduls *Machine Learning* an der OST – Ostschweizer Fachhochschule – durchgeführt. Ziel des Projekts ist es, eine datengetriebene **Relative-Value-Trading-Strategie** zu entwickeln und zu evaluieren. Dabei wird mithilfe von **XGBoost**, einem leistungsfähigen Gradient-Boosting-Algorithmus, versucht, profitable Handelsentscheidungen auf Basis historischer Kursdaten zu treffen. 

Das Projekt dient als praktische Vertiefung in Machine-Learning-Methoden im Finanzkontext und legt besonderen Fokus auf die Verbindung zwischen quantitativer Analyse. Der gesamte Prozess von der Datensammlung über die explorative Datenanalyse (EDA) bis hin zum Aufbau eines Modells wird dabei systematisch dokumentiert.

## 1.2. Motivation

Die Motivation für dieses Projekt ergibt sich aus der thematischen Nähe zu unserer geplanten **Bachelorarbeit**, in der wir uns mit *Statistical Arbitrage* unter Verwendung von *Topological Deep Learning* beschäftigen. Während dort komplexere Strukturen aus mehreren Aktien in einem Netzwerk betrachtet werden, fokussiert sich dieses Projekt auf klassische **Pairs-Trading-Strategien**, bei denen immer zwei Aktien miteinander verglichen werden. 

Das aktuelle Projekt dient somit als vereinfachte, aber praxisnahe Vorstufe zur Bachelorarbeit. Es erlaubt uns, den vollständigen Machine-Learning-Workflow im Kontext von Finanzdaten zu durchlaufen und gleichzeitig ein tieferes Verständnis für datengetriebene Arbitragestrategien zu entwickeln.

## 1.4. Theoriehintergrund (Kurzüberblick)

- **Pairs Trading:**
  Pairs Trading ist eine marktneutrale Handelsstrategie, bei der zwei historisch korrelierte/kointegrierte Aktien beobachtet werden. Weichen ihre Preisverhältnisse signifikant voneinander ab, wird erwartet, dass sie sich langfristig wieder annähern. Dies ermöglicht den Aufbau von Long- und Short-Positionen zur Ausnutzung dieser temporären Marktineffizienzen.

- **XGBoost:**
  XGBoost (*Extreme Gradient Boosting*) ist ein ensemblebasierter Machine-Learning-Algorithmus, der auf Entscheidungsbäumen basiert. Er zeichnet sich durch hohe Effizienz, Robustheit gegenüber Overfitting und hervorragende Performance in vielen Klassifikations- und Regressionsaufgaben aus. Für finanzielle Zeitreihen eignet er sich insbesondere, da er komplexe, nichtlineare Zusammenhänge zwischen Features erkennen kann.
